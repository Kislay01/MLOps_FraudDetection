```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', 100)

train_tx = pd.read_csv('../data/raw/train_transaction.csv')
train_id = pd.read_csv('../data/raw/train_identity.csv')

df = train_tx.merge(train_id, on='TransactionID', how='left')

print(train_tx.shape, train_id.shape, df.shape)
df.head()
```

```python
fraud_counts = df['isFraud'].value_counts()
fraud_pct = df['isFraud'].value_counts(normalize=True) * 100

print(fraud_counts)
print(fraud_pct)
```

```python
missing = df.isnull().mean().sort_values(ascending=False) * 100
missing = missing[missing > 0]

print(f"Total columns with missing values: {len(missing)}")
print(missing.head(20))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,4))
missing.hist(bins=50)
plt.xlabel('% missing')
plt.ylabel('number of columns')
plt.title('Distribution of missingness across columns')
plt.show()

print(missing.describe())
print("\nColumns >90% missing:", (missing > 90).sum())
print("Columns >50% missing:", (missing > 50).sum())
print("Columns >30% missing:", (missing > 30).sum())
```

```python
cols_to_drop = missing[missing > 90].index.tolist()
print(f"Dropping {len(cols_to_drop)} columns:")
print(cols_to_drop)

df = df.drop(columns=cols_to_drop)
print(f"\nNew shape: {df.shape}")
```

```python
print(df['TransactionDT'].describe())
print("\nRange in days:", (df['TransactionDT'].max() - df['TransactionDT'].min()) / (24*60*60))

plt.figure(figsize=(10,4))
df['TransactionDT'].hist(bins=100)
plt.xlabel('TransactionDT (seconds from reference point)')
plt.ylabel('number of transactions')
plt.title('Transaction volume over time')
plt.show()
```

```python
print("Unique card1 values:", df['card1'].nunique())
print("\nTransactions per card1 (distribution):")
print(df['card1'].value_counts().describe())

top_cards = df['card1'].value_counts().head(10)
print("\nTop 10 most active cards:")
print(top_cards)
```

```python
categorical_cols = ['ProductCD', 'card4', 'card6', 'P_emaildomain', 'R_emaildomain', 'M1', 'M2', 'M3']

for col in categorical_cols:
    if col in df.columns:
        print(f"{col}: {df[col].nunique()} unique values")
        print(df[col].value_counts(dropna=False).head(5))
        print()
```

```python
print("P_emaildomain:", df['P_emaildomain'].nunique(), "unique values")
print(df['P_emaildomain'].value_counts(dropna=False).head(10))
print()
print("R_emaildomain:", df['R_emaildomain'].nunique(), "unique values")
print(df['R_emaildomain'].value_counts(dropna=False).head(10))
```

```python
plt.figure(figsize=(10,4))
df[df['isFraud']==0]['TransactionAmt'].apply(np.log1p).hist(bins=50, alpha=0.5, label='Not Fraud')
df[df['isFraud']==1]['TransactionAmt'].apply(np.log1p).hist(bins=50, alpha=0.5, label='Fraud')
plt.legend()
plt.xlabel('log(TransactionAmt + 1)')
plt.title('Transaction amount distribution by class')
plt.show()
```

```python
for col in ['ProductCD', 'card4', 'card6']:
    print(f"--- Fraud rate by {col} ---")
    print(df.groupby(col)['isFraud'].agg(['mean', 'count']).sort_values('mean', ascending=False))
    print()
```

```python
v_cols = [c for c in df.columns if c.startswith('V')]
print(f"Total V columns: {len(v_cols)}")

# Missingness pattern grouping - columns with identical missingness % often belong to the same "block"
v_missing = df[v_cols].isnull().mean().round(4)
print("\nUnique missingness percentages among V columns:", v_missing.nunique())
print(v_missing.value_counts().head(15))
```

```python
# Take the block of 46 columns with ~77.9% missingness as a test case
block_cols = v_missing[v_missing == 0.7791].index.tolist()
print(f"Testing block of {len(block_cols)} columns: {block_cols[:5]}...")

corr_matrix = df[block_cols].corr().abs()
# Average correlation excluding self-correlation (diagonal)
avg_corr = (corr_matrix.sum().sum() - len(block_cols)) / (len(block_cols)**2 - len(block_cols))
print(f"\nAverage pairwise correlation within this block: {avg_corr:.3f}")
```

```python
device_cols = ['DeviceType', 'DeviceInfo']
for col in device_cols:
    if col in df.columns:
        print(f"--- {col} ---")
        print(f"Missing: {df[col].isnull().mean()*100:.1f}%")
        print(df.groupby(col)['isFraud'].agg(['mean', 'count']).sort_values('count', ascending=False).head(10))
        print()
```

```python
df['day'] = df['TransactionDT'] // (24*60*60)

daily_fraud = df.groupby('day')['isFraud'].agg(['mean', 'count'])

plt.figure(figsize=(14,4))
plt.plot(daily_fraud.index, daily_fraud['mean'])
plt.xlabel('Day')
plt.ylabel('Fraud rate')
plt.title('Daily fraud rate over the 182-day period')
plt.axhline(y=df['isFraud'].mean(), color='r', linestyle='--', label='Overall average')
plt.legend()
plt.show()

print(daily_fraud['mean'].describe())
```
